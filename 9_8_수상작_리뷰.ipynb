( https://www.kaggle.com/code/michaelsemenoff/ensemble-starter )

**데이터 설명**:
- 원본 데이터로부터 다양한 feature 엔지니어링을 수행하고

- 결과를 .pkl 형태로 저장한다

- 이후 모델 학습 및 앙상블에 활용될 준비된 데이터셋 생성 목적

#**[코드 리뷰]**
**1. EDA (탐색적 데이터 분석)**
- 데이터의 분포, 결측, 이상치, feature 상관관계 등을 심도 있게 분석
- EDA가 효율적으로 수행되지 않으면 .pkl 저장까지 이어지는 복잡한 파이프라인이 쉽게 지연될 수 있음

---

- 결측 처리

- 새로운 feature 생성

- 데이터 정제


---
**3. 모델링 (Modeling)**
- 여러 기본 모델을 학습시키고, 이들의 예측을 통합하는 앙상블 기법 활용

#**[차별 점 및 배울 점]**

✅ 차별 점
- 데이터 준비 자동화 및 재현성: 복잡한 전처리 과정을 자동화하고 .pkl 파일로 저장해 두는 패턴은 실무와 대회 환경 모두에서 매우 유용합니다. 반복적 실행 없이 바로 모델링 단계로 넘어갈 수 있죠.

- 효율성 중심 접근: 전처리–저장–학습으로 이어지는 구조는 대규모 데이터를 다룰 때 처리 시간을 크게 줄여 줍니다.


✨ 배울 점
- 전처리와 저장의 분리: 초기 노트북 내에서 전처리 로직과 모델링 로직을 분리하는 구조를 통해, 실험 반복 시 효율성을 확보하는 방법을 배울 수 있습니다.

- 재활용 가능한 데이터 파이프라인 구축: .pkl로 준비된 데이터를 여러 모델 또는 앙상블 방식에 쉽게 재사용할 수 있는 유연한 구조를 설계하는 것이 중요합니다.
